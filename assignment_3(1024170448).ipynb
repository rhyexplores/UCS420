```python
import pandas as pd

# Creating the data dictionary matching the table
data = {
    'Employee_ID':[101,102,103,104,105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Edward'],
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'Sales'],
    'Age':[29,34,41,28,38],
    'Salary':[2020-3-15,2017-7-19,2013-6-1,2021-2-10,2010-11-25],
    'Years_of_Experience':[4,8,10,3,12],
    'Joining_Date': ['2020-03-15', '2017-07-19', '2013-06-01', '2021-02-10', '2010-11-25'],
    'Gender': ['Female', 'Male', 'Male', 'Female', 'Male'],
    'Bonus':[5000,7000,6000,4500,5000],
    'Rating': [4.3, 4.0, 3.8, 4.7, 3.5]
}
dd=pd.DataFrame(data)
dd.to_csv('employees.csv',index=False)
print("humne is data ko dd.tocsv() se csv file bna di ")
```

```python
df=pd.read_csv("employees.csv")
print("Shape: ",df.shape)
print("Summary: ",df.info())
print("~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~")
print("Descriptive stats: ",df.describe())
print(df.head(),df.tail(3))
print("~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~")
print("Average salary: ",df['Salary'].mean())
print("Total bonus: ",df['Bonus'].sum())
print("Youngest employee: ",df['Age'].min())
print("Highest rating: ",df['Rating'].max())
print("~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~")
df=df.sort_values(by='Salary',ascending=False)
print(df.head(3))#df has sort_values func jisme ye do parameters hote hain
print("~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~")
liking=[]
for i in df['Rating']:
    if i>=4.5:
        liking.append('excellent')
    elif i>=4.0:
        liking.append('good')
    else:
        liking.append('average')
df['like_level']=liking
print(df[['Name','Rating','like_level']])
print("~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~")
print(df.isnull())
print(df.isnull().sum())
df=df.rename(columns={'Employee_ID':'ID'})
print(df.columns)
print(df[df['Years_of_Experience']>5])
print(df[df['Department']=='IT'])
df['Tax']=df['Salary']*0.10
df.to_csv('newcsv.csv',index='False')
```
