```python
import numpy as np
q1=np.array([10,100,1000,10000,100000])
print(q1+2)
print(q1*3)
print(q1/5)
```

```python
arr = np.array([1, 2, 3, 6, 4, 5])
print(arr[::-1])
x = np.array([1,2,3,4,5,1,2,1,1,1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])
xfreq=np.bincount(x).argmax()
xi=np.where(x==xfreq)
yfreq=np.bincount(y).argmax()
yi=np.where(y==yfreq)
print(xfreq,"pos",xi)
print(yfreq,"pos",yi)# 1 and 3 both occur same times but one comes/occurs first
```

```python
rhythm=np.linspace(10,100,25)
print(rhythm)
print(rhythm.shape)
print(rhythm.size)
print(rhythm.ndim)
print(rhythm.dtype)
print(rhythm.nbytes)
print(rhythm.reshape(-1,1))
print(rhythm.T)#this requires two dimensions to swap
```

```python
ucs420_rhythm = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])
print(ucs420_rhythm.shape)
print(np.mean(ucs420_rhythm))
print(np.median(ucs420_rhythm))
print(np.min(ucs420_rhythm))
print(np.max(ucs420_rhythm))
print(np.unique(ucs420_rhythm))
reshaped_ucs420_rhythm=ucs420_rhythm.reshape(4,3)
print(reshaped_ucs420_rhythm)
resized_ucs420_rhythm=np.resize(ucs420_rhythm,(2,3))
print(resized_ucs420_rhythm)
resized_ucs420_rhythm=np.resize(ucs420_rhythm,(2,8))
print(resized_ucs420_rhythm)
```
