```python
roll_no=[1,0,2,4,1,7,0,4,4,8]
rn=[digit*10 for digit in roll_no]
print(rn)

rn.insert(4,77)
print(rn)
rn.append(88)
print(rn)

rn.remove(10)
print(rn)
rn.pop(3)
print(rn)

rn.sort()
print(rn)
#rn.reverse()
rn.sort(reverse=True)
print(rn)

print(rn[:3])
print(rn[-3:])

avg=sum(rn)/len(rn)
rnn=[newrn for newrn in rn if newrn>avg]
print(avg," : ",rnn)
```

```python
q2=tuple(rnn[:8])
print(q2)
maxx=max(q2)
print(q2.index(maxx))
minn=min(q2)
print(q2.count(minn))

guess=int(input("enter the number: "))
if guess in q2:
    print("score found")
else:
    print("score not found")

try:
    q2[2]=300
except TypeError as e:
    print(e," tuple is immutable we cant change whereas list is mutable")
s1,s2,*rem=q2
print("pehla: ",s1," dusra: ",s2," remaining: ",rem)
```

```python
import random as ran
ran.seed(1024170448)
n1=[ran.randint(100,900) for _ in range(100)]
en=[n for n in n1 if n%2==0]
on=[n for n in n1 if n%2!=0]
print(len(en))
print(len(on))

def isprime(n):
    if n<2:
        return False
    for i in range(2,int(n*0.5)+1):
        if n%i==0:
            return False
    return True
pn=[no for no in n1 if isprime(no)]
print(pn)
f=max(set(n1),key=n1.count)
ff=n1.count(f)
print(ff,": ",f)
```

```python
q4=[1,0,2,4,1,7,0,4]
set1={no*7 for no in q4}
set2={no*9 for no in q4}
print(set1,"  ",set2)

print(set1|set2)
print(set1&set2)
print(set1-set2)
print(set2.difference(set1))

print(set1.symmetric_difference(set2))
print(set1.issubset(set2))
print(set2.issuperset(set1))

print("before: ",set1)
inn=int(input("enter the number: "))
set1.discard(inn)#discard doesnt raise key error like remove if number not existed
print(inn," -->",set1)
```
