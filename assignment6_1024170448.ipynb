```python
import numpy as np
# 4 users (rows) x 3 days (columns)
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])
total_steps = np.sum(steps)
print(f"a. Total steps recorded: {total_steps}")
mean_steps = np.mean(steps)
print(f"b. Mean number of steps: {mean_steps:.2f}")
max_steps = np.max(steps)
min_steps = np.min(steps)
print(f"c. Maximum steps: {max_steps}, Minimum steps: {min_steps}")
daily_totals = np.sum(steps, axis=0)
print(f"d. Total steps for each day: {daily_totals}")
user_totals = np.sum(steps, axis=1)
print(f"e. Total steps for each user: {user_totals}")
# argmax() on a 2D array returns a flat index, np.unravel_index converts it to (row, col)
flat_max_index = np.argmax(steps)
user_idx, day_idx = np.unravel_index(flat_max_index, steps.shape)
print(f"f. Maximum steps position: User {user_idx + 1}, Day {day_idx + 1} (Indices: row {user_idx}, col {day_idx})")
```

```python
import numpy as np
X = np.array([ [6, 70, 3], [5, 50, 6], [8, 80, 2], [4, 30, 8] ])
y = np.array([40, 65, 30, 85])
print("shape: ",X.shape)
print("shape: ",X.shape)
# 𝛽 = (𝑋^t.𝑋)^-1 * 𝑋^t * 𝑦 using NumPy
x_ka_transpose=X.T
print(x_ka_transpose.shape)
prod=x_ka_transpose@X
print(prod)
prodinv=np.linalg.inv(prod)
beta=prodinv@x_ka_transpose@y
print(beta)
#Coefficients: estimated change in the user's assistance score for a one-unit increase in that specific #feature (Sleep hours, Activity level, or Stress level), assuming all other features remain constant.
new_user = np.array([5, 40, 7])
pred=new_user@beta
print(pred)
```
